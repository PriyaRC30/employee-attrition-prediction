```python
# ============================================================
# EMPLOYEE ATTRITION PREDICTION
# EDA + DATA PREPROCESSING
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------

import os
import joblib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

print("=" * 60)
print("EMPLOYEE ATTRITION PREDICTION")
print("EDA AND DATA PREPROCESSING")
print("=" * 60)


# -----------------------------
# 2. PROBLEM IDENTIFICATION
# -----------------------------

print("\nPROBLEM STATEMENT")
print("-" * 40)
print("""
Employee attrition is an important problem for organizations.
This project aims to predict whether an employee is likely to
leave the organization using demographic, job, salary,
satisfaction and work-related features.
""")

print("OBJECTIVE")
print("-" * 40)
print("""
Build a machine learning classification system that predicts
employee attrition and helps organizations identify employees
who may be at risk of leaving.
""")


# -----------------------------
# 3. LOAD DATASET
# -----------------------------

data_path = "../data/WA_Fn-UseC_-HR-Employee-Attrition.csv"

if not os.path.exists(data_path):
    raise FileNotFoundError(
        f"Dataset not found at: {data_path}\n"
        "Check that the CSV file is inside the data folder."
    )

df = pd.read_csv(data_path)

print("\nDATASET LOADED SUCCESSFULLY")
print("-" * 40)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nFirst 5 rows:")
display(df.head())


# -----------------------------
# 4. DATASET INFORMATION
# -----------------------------

print("\nDATASET INFORMATION")
print("-" * 40)

print("Shape:", df.shape)

print("\nData types:")
print(df.dtypes)


# -----------------------------
# 5. MISSING VALUES
# -----------------------------

print("\nMISSING VALUE ANALYSIS")
print("-" * 40)

missing_values = df.isnull().sum()

print("Total missing values:", missing_values.sum())

if missing_values.sum() == 0:
    print("No missing values found.")
else:
    display(missing_values[missing_values > 0])


# -----------------------------
# 6. DUPLICATES
# -----------------------------

print("\nDUPLICATE ANALYSIS")
print("-" * 40)

duplicates = df.duplicated().sum()

print("Duplicate rows:", duplicates)


# -----------------------------
# 7. TARGET VARIABLE
# -----------------------------

print("\nTARGET VARIABLE: ATTRITION")
print("-" * 40)

print("Unique values:")
print(df["Attrition"].unique())

attrition_counts = df["Attrition"].value_counts()

attrition_percentages = (
    df["Attrition"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

attrition_summary = pd.DataFrame({
    "Count": attrition_counts,
    "Percentage": attrition_percentages
})

display(attrition_summary)


# -----------------------------
# 8. TARGET VISUALIZATION
# -----------------------------

plt.figure(figsize=(7, 5))

ax = sns.countplot(
    data=df,
    x="Attrition",
    hue="Attrition",
    palette={"No": "#2E86DE", "Yes": "#E74C3C"},
    legend=False
)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

total = len(df)

for container in ax.containers:
    labels = []

    for bar in container:
        count = int(bar.get_height())
        percentage = count / total * 100
        labels.append(f"{count} ({percentage:.1f}%)")

    ax.bar_label(container, labels=labels, padding=3)

plt.tight_layout()
plt.show()


# -----------------------------
# 9. IDENTIFY CONSTANT COLUMNS
# -----------------------------

print("\nCONSTANT COLUMN ANALYSIS")
print("-" * 40)

unique_counts = df.nunique()

constant_columns = unique_counts[
    unique_counts == 1
].index.tolist()

print("Constant columns:")
print(constant_columns)


# -----------------------------
# 10. COLUMNS TO REMOVE
# -----------------------------

columns_to_drop = [
    "EmployeeCount",
    "EmployeeNumber",
    "Over18",
    "StandardHours"
]

print("\nCOLUMNS REMOVED FROM MODEL")
print("-" * 40)

for column in columns_to_drop:
    print(column)


# -----------------------------
# 11. SEPARATE FEATURES AND TARGET
# -----------------------------

X = df.drop(columns=["Attrition"])

y = df["Attrition"].map({
    "No": 0,
    "Yes": 1
})

print("\nFEATURE AND TARGET")
print("-" * 40)

print("Feature shape:", X.shape)
print("Target shape:", y.shape)


# ============================================================
# EDA
# ============================================================

print("\n")
print("=" * 60)
print("EXPLORATORY DATA ANALYSIS")
print("=" * 60)


# -----------------------------
# 12. AGE VS ATTRITION
# -----------------------------

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="Attrition",
    y="Age",
    hue="Attrition",
    palette={"No": "#2E86DE", "Yes": "#E74C3C"},
    legend=False
)

plt.title("Age Distribution by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Age")

plt.tight_layout()
plt.show()


# -----------------------------
# 13. MONTHLY INCOME VS ATTRITION
# -----------------------------

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="Attrition",
    y="MonthlyIncome",
    hue="Attrition",
    palette={"No": "#2E86DE", "Yes": "#E74C3C"},
    legend=False
)

plt.title("Monthly Income by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")

plt.tight_layout()
plt.show()


# -----------------------------
# 14. OVERTIME VS ATTRITION
# -----------------------------

overtime_rate = pd.crosstab(
    df["OverTime"],
    df["Attrition"],
    normalize="index"
) * 100

print("\nAttrition percentage by Overtime:")
display(overtime_rate.round(2))

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="OverTime",
    hue="Attrition",
    palette={"No": "#2E86DE", "Yes": "#E74C3C"}
)

plt.title("Employee Attrition by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()


# -----------------------------
# 15. JOB ROLE VS ATTRITION
# -----------------------------

jobrole_rate = pd.crosstab(
    df["JobRole"],
    df["Attrition"],
    normalize="index"
) * 100

print("\nAttrition percentage by Job Role:")
display(jobrole_rate.round(2))

plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    x="JobRole",
    hue="Attrition",
    palette={"No": "#2E86DE", "Yes": "#E74C3C"}
)

plt.title("Employee Attrition by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()


# -----------------------------
# 16. JOB SATISFACTION VS ATTRITION
# -----------------------------

satisfaction_rate = pd.crosstab(
    df["JobSatisfaction"],
    df["Attrition"],
    normalize="index"
) * 100

print("\nAttrition percentage by Job Satisfaction:")
display(satisfaction_rate.round(2))

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="JobSatisfaction",
    hue="Attrition",
    palette={"No": "#2E86DE", "Yes": "#E74C3C"}
)

plt.title("Employee Attrition by Job Satisfaction")
plt.xlabel("Job Satisfaction Level")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()


# -----------------------------
# 17. YEARS AT COMPANY VS ATTRITION
# -----------------------------

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="Attrition",
    y="YearsAtCompany",
    hue="Attrition",
    palette={"No": "#2E86DE", "Yes": "#E74C3C"},
    legend=False
)

plt.title("Years at Company by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Years at Company")

plt.tight_layout()
plt.show()


# -----------------------------
# 18. CORRELATION HEATMAP
# -----------------------------

numeric_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

correlation_matrix = df[numeric_columns].corr()

plt.figure(figsize=(15, 11))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0,
    linewidths=0.3
)

plt.title("Correlation Heatmap of Numerical Features")

plt.tight_layout()
plt.show()


# ============================================================
# DATA PREPROCESSING
# ============================================================

print("\n")
print("=" * 60)
print("DATA PREPROCESSING")
print("=" * 60)


# -----------------------------
# 19. TRAIN TEST SPLIT
# -----------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("\nTRAIN TEST SPLIT")
print("-" * 40)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)


# -----------------------------
# 20. REMOVE IRRELEVANT COLUMNS
# -----------------------------

X_train_clean = X_train.drop(
    columns=columns_to_drop,
    errors="ignore"
)

X_test_clean = X_test.drop(
    columns=columns_to_drop,
    errors="ignore"
)

print("\nAfter removing irrelevant columns:")
print("Training:", X_train_clean.shape)
print("Testing:", X_test_clean.shape)


# -----------------------------
# 21. IDENTIFY FEATURE TYPES
# -----------------------------

numeric_features = X_train_clean.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train_clean.select_dtypes(
    include=["object"]
).columns.tolist()

print("\nNumerical features:", len(numeric_features))
print(numeric_features)

print("\nCategorical features:", len(categorical_features))
print(categorical_features)


# -----------------------------
# 22. CREATE PREPROCESSING PIPELINE
# -----------------------------

numeric_transformer = Pipeline(
    steps=[
        ("scaler", StandardScaler())
    ]
)

categorical_transformer = Pipeline(
    steps=[
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)


# -----------------------------
# 23. FIT PREPROCESSOR
# -----------------------------

preprocessor.fit(X_train_clean)

print("\nPreprocessing pipeline fitted successfully.")


# -----------------------------
# 24. TRANSFORM DATA
# -----------------------------

X_train_processed = preprocessor.transform(
    X_train_clean
)

X_test_processed = preprocessor.transform(
    X_test_clean
)

print("\nProcessed training shape:",
      X_train_processed.shape)

print("Processed testing shape:",
      X_test_processed.shape)


# -----------------------------
# 25. CHECK PROCESSED DATA
# -----------------------------

feature_names = preprocessor.get_feature_names_out()

X_train_processed_df = pd.DataFrame(
    X_train_processed,
    columns=feature_names
)

X_test_processed_df = pd.DataFrame(
    X_test_processed,
    columns=feature_names
)

print("\nProcessed data sample:")
display(X_train_processed_df.head())

print(
    "Missing values after preprocessing:",
    X_train_processed_df.isnull().sum().sum()
)


# -----------------------------
# 26. SAVE PREPROCESSING PIPELINE
# -----------------------------

os.makedirs("../models", exist_ok=True)

pipeline_path = "../models/preprocessing_pipeline.pkl"

joblib.dump(
    preprocessor,
    pipeline_path
)

print("\nPreprocessing pipeline saved:")
print(pipeline_path)


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 60)
print("PROJECT PHASE 1 & 2 COMPLETED")
print("=" * 60)

print("""
Completed:
1. Problem identification
2. Dataset loading
3. Dataset inspection
4. Missing-value analysis
5. Duplicate analysis
6. Target-variable analysis
7. EDA and visualization
8. Irrelevant-column removal
9. Train-test split
10. Numerical feature scaling
11. Categorical feature encoding
12. Preprocessing pipeline creation
13. Preprocessing pipeline saved

The dataset is now ready for machine learning model training.
""")
```
